<h3>Setting Up Resource Monitor</h3>
Go to Admin
Select Cost Management
Click Resource Monitors tab
Click +Resource Monitor

New Resource Monitor Page opens
Name: MonthlyAccountBudget
Credit Quota: 10
Monitor Type: Account or Warehouse
Schedule: Starts monitoring Immediately
          End monitoring Never
          Resets Monthly
Actions: 99% Suspend Immediately and Notify when this % of credit is used
         80% Suspend and Notify when this % of credit is used
         50% Notify when this % of credit is used
Click Create Resource Monitor
